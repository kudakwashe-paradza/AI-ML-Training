# House Prices — One-Hot Encoding
This notebook follows **Method 2 (One-Hot Encoding)** from the encoding-techniques
reference notebook, applied to the Ames Housing dataset (`train.csv` / `test.csv`,
target = `SalePrice`).

Steps:
1. Load data
2. Split features / target
3. Handle missing values
4. One-hot encode categorical columns (`pd.get_dummies`), aligning train/test columns
5. Train a model (Linear Regression + Random Forest) and evaluate
6. Predict on the test set and build a submission file

## 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.impute import SimpleImputer
```

## 2. Load the data

```python
df_train = pd.read_csv('train.csv')
df_test = pd.read_csv('test.csv')

print('train data set has got {} rows and {} columns'.format(df_train.shape[0], df_train.shape[1]))
print('test data set has got {} rows and {} columns'.format(df_test.shape[0], df_test.shape[1]))

df_train.head()
```

## 3. Define train and target

```python
X = df_train.drop(['SalePrice', 'Id'], axis=1)
y = df_train['SalePrice']

test_ids = df_test['Id']
X_test_final = df_test.drop(['Id'], axis=1)

X['__is_train__'] = 1
X_test_final['__is_train__'] = 0
combined = pd.concat([X, X_test_final], axis=0, sort=False)

print('combined shape:', combined.shape)
```

## 4. Handle missing values
Many of the "missing" values here aren't really missing — for categorical columns like
`PoolQC`, `Alley`, `Fence`, `GarageType`, a NaN means the house simply doesn't have that
feature. We fill those with the string `'None'` instead of dropping them. Numeric columns
are filled with the median.

```python
cat_cols = combined.select_dtypes(include='object').columns.tolist()
num_cols = combined.select_dtypes(include=['int64', 'float64']).columns.tolist()
num_cols.remove('__is_train__')

print('categorical columns:', len(cat_cols))
print('numeric columns:', len(num_cols))

combined[cat_cols] = combined[cat_cols].fillna('None')

train_mask = combined['__is_train__'] == 1
for c in num_cols:
    median_val = combined.loc[train_mask, c].median()
    combined[c] = combined[c].fillna(median_val)

print('remaining NaNs:', combined.isna().sum().sum())
```

## 5. Method 2 — One-Hot Encoding
Using `pd.get_dummies()`, exactly like Method 2 in the reference notebook. Because we
encoded the **combined** train+test frame, both sides end up with identical dummy
columns — no column-mismatch issues when we split them back apart.

```python
%%time

combined_encoded = pd.get_dummies(combined, columns=cat_cols, drop_first=True)

print('encoded data set has got {} rows and {} columns'.format(
    combined_encoded.shape[0], combined_encoded.shape[1]))
combined_encoded.head()
```

```python
train_encoded = combined_encoded[combined_encoded['__is_train__'] == 1].drop('__is_train__', axis=1)
test_encoded = combined_encoded[combined_encoded['__is_train__'] == 0].drop('__is_train__', axis=1)

print('train_encoded:', train_encoded.shape)
print('test_encoded:', test_encoded.shape)
```

## 6. Train / validation split

```python
X_train, X_val, y_train, y_val = train_test_split(
    train_encoded, y, random_state=42, test_size=0.2
)

print(X_train.shape, X_val.shape)
```

## 7. Train models
`SalePrice` is a continuous target, so this is regression, not classification —
we use RMSE and R² instead of accuracy.

```python
def evaluate(model, X_train, y_train, X_val, y_val, name):
    model.fit(X_train, y_train)
    preds = model.predict(X_val)
    rmse = np.sqrt(mean_squared_error(y_val, preds))
    r2 = r2_score(y_val, preds)
    print(f'{name:>20s}  RMSE: {rmse:,.0f}   R2: {r2:.4f}')
    return model, rmse, r2

results = {}

lr = LinearRegression()
_, rmse_lr, r2_lr = evaluate(lr, X_train, y_train, X_val, y_val, 'Linear Regression')
results['Linear Regression (OHE)'] = (rmse_lr, r2_lr)

rf = RandomForestRegressor(n_estimators=300, random_state=42, n_jobs=-1)
_, rmse_rf, r2_rf = evaluate(rf, X_train, y_train, X_val, y_val, 'Random Forest')
results['Random Forest (OHE)'] = (rmse_rf, r2_rf)
```

## 8. Summary of model performance

```python
summary = pd.DataFrame(results, index=['RMSE', 'R2']).T
summary
```

## 9. Predict on the test set & build a submission
Refit the better model on **all** of the training data, then predict on the
(one-hot encoded) Kaggle test set.

```python
best_model = rf if rmse_rf < rmse_lr else lr
best_model.fit(train_encoded, y)

test_preds = best_model.predict(test_encoded)

submission = pd.DataFrame({'Id': test_ids, 'SalePrice': test_preds})
submission.to_csv('submission_one_hot.csv', index=False)
submission.head()
```

### Notes / next steps
- `drop_first=True` avoids the dummy-variable trap (one redundant column per category) —
  drop it if your model doesn't care (e.g. tree-based models) or keep it for linear models.
- One-hot encoding here roughly matches the reference notebook's finding: it's a strong
  baseline, but on a dataset with many *high-cardinality* categorical columns
  (e.g. `Neighborhood`, `Exterior1st`) it inflates the column count a lot — Feature Hashing
  or Target Encoding (Methods 3 & 5 in the reference notebook) are worth comparing if you
  want to control dimensionality.
- Ordinal-feeling columns (`ExterQual`, `KitchenQual`, `PoolQC`, etc. which are really
  Excellent > Good > ... > Poor) are one-hot encoded here for consistency with Method 2,
  but you may get a bump by label/ordinal-encoding those specifically instead.
